```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
access_df = df[df['pillar'] == 'access'].sort_values('observation_date').copy()
access_df['observation_date'] = pd.to_datetime(access_df['observation_date'])
access_df['year'] = access_df['observation_date'].dt.year

access_values = access_df['adjusted_value'] if 'adjusted_value' in access_df.columns else access_df['value_numeric']
X_access = access_df['year'].values.reshape(-1,1)
y_access = access_values.values
model_access = LinearRegression()
model_access.fit(X_access, y_access)

future_years = np.array([2025, 2026, 2027]).reshape(-1,1)
access_forecast = model_access.predict(future_years)

usage_df = df[df['pillar'] == 'usage'].sort_values('observation_date').copy()
usage_df['observation_date'] = pd.to_datetime(usage_df['observation_date'])
usage_df['year'] = usage_df['observation_date'].dt.year

usage_values = usage_df['adjusted_value'] if 'adjusted_value' in usage_df.columns else usage_df['value_numeric']

X_usage = usage_df['year'].values.reshape(-1,1)
y_usage = usage_values.values
model_usage = LinearRegression()
model_usage.fit(X_usage, y_usage)
```

```python
usage_forecast = model_usage.predict(future_years)
```

```python
forecast_df = pd.DataFrame({
    'year': [2025, 2026, 2027],
    'Access_forecast (%)': access_forecast,
    'Usage_forecast (%)': usage_forecast
})
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(x='year', y='value_numeric', data=access_df, marker='o', label='Access Observed')
sns.lineplot(x='year', y='Access_forecast (%)', data=forecast_df, marker='o', label='Access Forecast')

sns.lineplot(x='year', y='value_numeric', data=usage_df, marker='o', label='Usage Observed')
sns.lineplot(x='year', y='Usage_forecast (%)', data=forecast_df, marker='o', label='Usage Forecast')

plt.title("Financial Inclusion Forecast (Access & Usage) — 2025 to 2027")
plt.ylabel("Percentage (%)")
plt.xlabel("Year")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
forecast_df['Access_Optimistic'] = forecast_df['Access_forecast (%)'] * 1.05
forecast_df['Access_Pessimistic'] = forecast_df['Access_forecast (%)'] * 0.95

forecast_df['Usage_Optimistic'] = forecast_df['Usage_forecast (%)'] * 1.05
forecast_df['Usage_Pessimistic'] = forecast_df['Usage_forecast (%)'] * 0.95

print("Forecast with scenarios (2025-2027):")
print(forecast_df)
```
